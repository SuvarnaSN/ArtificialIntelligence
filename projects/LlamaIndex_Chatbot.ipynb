```python
import os           # saves API key to the env variable
from llama_index.core.llms import ChatMessage, MessageRole
from llama_index.llms.groq import Groq
```

```python
# API KEY
import os
os.environ["GROQ_API_KEY"] = ""
```

```python
# Initiating LLMs
llm = Groq(
    model="llama-3.3-70b-versatile",
    temperature=0
)
```

```python
# chathistory is a list of messages and each message will have a role and content.
# role 1. system, 2.assistant
# thsi is passed to the chat prompt template so we just have to mention the tuples here.
# we will send the entire chat history to the LLM to give it proper context.

def chat():
  # Chatmessage object has a role and content
  history = [
      ChatMessage(role=MessageRole.SYSTEM, content="You are a helpful assistant. Be concise and accurate")
  ]

  print("LlamaIndex Chatbot. type 'exit' to quit.\n")

  # We need a while loop to keep asking the user to enter a question.
  while True:
    user = input("You: ").strip()
    if user.lower() == 'exit':
      break;

    # add user message to the chat history for context. This is of the form role, context
    # Both the role and context should be in the form of a tuple.
    history.append(ChatMessage(role=MessageRole.USER, content=user))

    # call LLM with full history so that LLM can look at the history
    resp = llm.chat(messages=history)
    answer = resp.message.content

    print(f"Bot: {answer}")

    # save the bot message in chat history
    history.append(ChatMessage(role=MessageRole.ASSISTANT, content=answer))
    print("-"*80)
```

```python
chat()
```
